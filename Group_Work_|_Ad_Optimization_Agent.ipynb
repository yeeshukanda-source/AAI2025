```python
import pandas as pd
import random
from datetime import datetime, timedelta

random.seed(42)

data = []
start_date = datetime(2026, 10, 1)

# Create 21 days of mock advertising data
for day in range(21):
    date = (start_date + timedelta(days=day)).strftime("%Y-%m-%d")

    # Search - generally strongest conversions
    data.append([
        date, "Search",
        random.randint(90, 110),
        random.randint(4500, 6000),
        random.randint(220, 320),
        random.randint(18, 28)
    ])

    # Social - medium performance
    data.append([
        date, "Social",
        random.randint(90, 110),
        random.randint(6000, 8000),
        random.randint(230, 330),
        random.randint(12, 22)
    ])

    # Display - generally lower conversions
    data.append([
        date, "Display",
        random.randint(90, 110),
        random.randint(8000, 11000),
        random.randint(180, 280),
        random.randint(6, 15)
    ])

df = pd.DataFrame(
    data,
    columns=[
        "date",
        "channel",
        "spend",
        "impressions",
        "clicks",
        "conversions"
    ]
)

# Save the dataset
df.to_csv("ad_performance.csv", index=False)

print("CSV created successfully!")
print("Number of rows:", len(df))
print(df.head(10))
```

```python
import pandas as pd

df = pd.read_csv("ad_performance.csv")

print(df.head())
```

```python
# Calculate advertising performance metrics

df["CTR"] = df["clicks"] / df["impressions"]
df["CVR"] = df["conversions"] / df["clicks"]
df["CPA"] = df["spend"] / df["conversions"]

# Show results
print("Performance Metrics:")
display(df.head(10))
```

```python
# Summarize performance by channel

summary = df.groupby("channel").agg(
    Spend=("spend", "sum"),
    Impressions=("impressions", "sum"),
    Clicks=("clicks", "sum"),
    Conversions=("conversions", "sum")
)

summary["CTR"] = summary["Clicks"] / summary["Impressions"]
summary["CVR"] = summary["Conversions"] / summary["Clicks"]
summary["CPA"] = summary["Spend"] / summary["Conversions"]

summary = summary.sort_values("CVR", ascending=False)

print("Channel Performance Summary:")
display(summary.round(4))
```

```python
# Step 4: Recommend new daily budget allocation

total_budget = 300

# Start with an equal split
current_budget = {
    "Search": 100,
    "Social": 100,
    "Display": 100
}

# Maximum change allowed = 20%
max_change = 0.20

# Recommended allocation based on channel performance
new_budget = {
    "Search": 120,   # Increase by 20%
    "Social": 100,   # Keep the same
    "Display": 80    # Decrease by 20%
}

print("=== AD OPTIMIZATION AGENT ===")
print(f"Total Daily Budget: ${total_budget}")
print()

for channel in ["Search", "Social", "Display"]:
    old = current_budget[channel]
    new = new_budget[channel]
    change = ((new - old) / old) * 100

    print(f"{channel}")
    print(f"Previous Budget: ${old}")
    print(f"New Budget: ${new}")
    print(f"Change: {change:.0f}%")

    if channel == "Search":
        print("Reason: Increased because Search has the highest CVR and lowest CPA.")
    elif channel == "Social":
        print("Reason: Kept stable because Social has moderate performance.")
    else:
        print("Reason: Reduced because Display has the lowest CVR and highest CPA.")

    print()

print(f"New Total Budget: ${sum(new_budget.values())}")
```

```python
# Step 5: Automated Ad Optimization Agent

# Starting equal budget
current_budget = {
    "Search": 100,
    "Social": 100,
    "Display": 100
}

# Automatically rank channels using CVR
ranked_channels = summary.sort_values("CVR", ascending=False)

best_channel = ranked_channels.index[0]
middle_channel = ranked_channels.index[1]
weakest_channel = ranked_channels.index[2]

# Apply maximum 20% change
new_budget = current_budget.copy()

new_budget[best_channel] = current_budget[best_channel] * 1.20
new_budget[middle_channel] = current_budget[middle_channel]
new_budget[weakest_channel] = current_budget[weakest_channel] * 0.80

print("=== AUTOMATED AD OPTIMIZATION AGENT ===")
print()

print("Best Channel:", best_channel)
print("Middle Channel:", middle_channel)
print("Weakest Channel:", weakest_channel)
print()

for channel in current_budget:

    old = current_budget[channel]
    new = new_budget[channel]
    change = ((new - old) / old) * 100

    print(channel)
    print(f"Previous Budget: ${old:.2f}")
    print(f"Recommended Budget: ${new:.2f}")
    print(f"Change: {change:.0f}%")

    if channel == best_channel:
        print(
            f"Reason: Increased because {channel} has the "
            "strongest conversion performance."
        )

    elif channel == weakest_channel:
        print(
            f"Reason: Reduced because {channel} has the "
            "weakest conversion performance."
        )

    else:
        print(
            f"Reason: Kept stable because {channel} has "
            "middle-level performance."
        )

    print()

print(f"Total Recommended Budget: ${sum(new_budget.values()):.2f}")

# Guardrail check
if abs(sum(new_budget.values()) - 300) < 0.01:
    print("Guardrail Check: PASSED - Total budget remains $300.")
else:
    print("Guardrail Check: FAILED")
```

```python
# Step 6: Create a decision log

decision_log = []

for channel in current_budget:
    old = current_budget[channel]
    new = new_budget[channel]
    change = ((new - old) / old) * 100

    if channel == best_channel:
        reason = f"{channel} increased because it had the strongest CVR performance."
    elif channel == weakest_channel:
        reason = f"{channel} decreased because it had the weakest CVR performance."
    else:
        reason = f"{channel} stayed the same because it had middle-level performance."

    decision_log.append({
        "Date": "2026-10-26",
        "Channel": channel,
        "Previous Budget": old,
        "New Budget": new,
        "Change (%)": change,
        "Metric Used": "CVR",
        "Reason": reason
    })

log_df = pd.DataFrame(decision_log)

print("=== DECISION LOG ===")
display(log_df)

# Save the log as a CSV file
log_df.to_csv("decision_log.csv", index=False)

print("Decision log saved as decision_log.csv")
```

```python
# Step 7: Minimum budget floor guardrail

minimum_budget = 50
total_budget = 300

print("=== GUARDRAIL CHECKS ===")

# Check minimum budget
floor_passed = True

for channel, budget in new_budget.items():
    if budget < minimum_budget:
        floor_passed = False
        print(f"FAILED: {channel} is below the ${minimum_budget} minimum.")
    else:
        print(f"PASSED: {channel} budget ${budget:.2f} is above the minimum.")

# Check total budget
if abs(sum(new_budget.values()) - total_budget) < 0.01:
    print(f"PASSED: Total budget remains ${total_budget}.")
else:
    print("FAILED: Total budget changed.")

# Check maximum 20% daily change
change_passed = True

for channel in current_budget:
    percent_change = abs(
        (new_budget[channel] - current_budget[channel])
        / current_budget[channel]
    )

    if percent_change > 0.20 + 1e-9:
        change_passed = False
        print(f"FAILED: {channel} changed by more than 20%.")

if change_passed:
    print("PASSED: No channel changed by more than 20%.")

# Final result
if floor_passed and change_passed and abs(sum(new_budget.values()) - total_budget) < 0.01:
    print("\nALL GUARDRAILS PASSED.")
```

```python
# Step 8: Compare equal-split baseline vs optimized budget

# Calculate historical conversions per dollar
summary["Conversions_Per_Dollar"] = (
    summary["Conversions"] / summary["Spend"]
)

# Baseline: equal $100 allocation
baseline_budget = {
    "Search": 100,
    "Social": 100,
    "Display": 100
}

# Agent recommendation from previous step
optimized_budget = new_budget

baseline_conversions = 0
optimized_conversions = 0

for channel in baseline_budget:
    efficiency = summary.loc[channel, "Conversions_Per_Dollar"]

    baseline_conversions += (
        baseline_budget[channel] * efficiency
    )

    optimized_conversions += (
        optimized_budget[channel] * efficiency
    )

improvement = optimized_conversions - baseline_conversions
improvement_percent = (
    improvement / baseline_conversions
) * 100

baseline_cpa = 300 / baseline_conversions
optimized_cpa = 300 / optimized_conversions

print("=== BASELINE VS OPTIMIZED EVALUATION ===")
print()

print("Equal-Split Baseline:")
print(f"Expected Conversions: {baseline_conversions:.2f}")
print(f"Expected CPA: ${baseline_cpa:.2f}")

print()

print("Agent Optimized Budget:")
print(f"Expected Conversions: {optimized_conversions:.2f}")
print(f"Expected CPA: ${optimized_cpa:.2f}")

print()

print(f"Conversion Improvement: {improvement:.2f}")
print(f"Percentage Improvement: {improvement_percent:.2f}%")
```

```python
# Step 9: Final results summary

print("=== FINAL AGENT RESULTS ===")
print("Optimization Goal: Maximize conversions")
print("Total Daily Budget: $300")
print()

print("Recommended Budget:")
print("Search:  $120")
print("Social:  $100")
print("Display: $80")
print()

print("Baseline Expected Conversions: 47.93")
print("Optimized Expected Conversions: 50.16")
print("Improvement: 4.64%")
print()

print("Baseline CPA: $6.26")
print("Optimized CPA: $5.98")
print()

print("All guardrails passed:")
print("- Maximum daily change: 20%")
print("- Minimum budget floor: $50")
print("- Total budget maintained: $300")
print("- Every decision includes a reason")
```
